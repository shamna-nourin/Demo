Constructor

```python
class Student:
  def __init__(self,name,age):
    self.name=name   #instance variable
    self.age=age     #instance variable
student1=Student("Abhinav",25)
print(student1.name)
print(student1.age)
```

```python
student2=Student("Anu",23)
print(student2.name)
print(student2.age)
```

Instance VS Class Variable:

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name       #instance variable
        self.salary = salary   #instance variable

    def display(self):
        print("Employee:", self.name)
        print("Salary:", self.salary)


emp1 = Employee("Arun", 30000)
emp2 = Employee("Anu", 40000)

emp1.display()
emp2.display()
```

```python
class Car:
  brand="Toyota"   #class variable
  def __init__(self,model):
    self.model=model
c1=Car("Hycross")
c2=Car("Crysta")
print(c1.brand)
print(c2.brand)
print(c1.model)
```

Inheritance:

```python
class Animal:
  def eat(self):
    print("Animal is eating")

class Dog(Animal):
  def bark(self):
    print("Dog is barking")

D=Dog()
print(D.eat())
D.bark()
```

Method Overiding:

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")

animal = Animal()
dog = Dog()

animal.sound()
dog.sound()
```

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")

dog = Dog()
dog.sound()
```

```python
d=Animal()
d.sound()
```

Encapsulation:

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance
account=BankAccount(50000)
# account.__balance()
print(account.get_balance())
```

Polymorphism:

```python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")
C=Cat()
D=Dog()
D.sound()
C.sound()
```

```python
class Student:
  institute="Entri Elevate"

  def __init__(self,name,age,course):
    self.name=name
    self.age=age
    self.course=course

  def display(self):
    print("Name: ",self.name)
    print("Age: ",self.age)
    print("Course: ",self.course)
    print("Institute: ",self.institute)

s1=Student("Rahul",23,"Data Analytics")
s2=Student("Teslin",21,"Data Science")

s1.display()
print("--------------------------------")
s2.display()
```
