```python
name = 'bot207'
weather = 'sunny'
mood = 'great'


responses = {
    'hello':['hy','hey'],
    "what's your name?" : ['They call me{}'.format(name),
                         'I usually go by the name {}'.format(name),
                         'I am {}'.format(name)],
    "what's today's weather?" : ['The weather is {}'.format(weather),
                                 "it's {} today".format(weather)],
    "how are you?" : ["I am feeling {}" .format(mood),
                      "{}! How about you?".format(mood),
                    "I am {}! How about you?".format(mood)],

                "" : ["Hey! Are you there?",
                   "What do you mean by these?"],

               "default" : ["Sorry!!! I didn't get you","Sorry??"],

                'end' : ["Thank you!!","Good to talk to you!!"]
}
```

```python
import random
def res(message):
    if message in responses:
        bot207_message = random.choice(responses[message])
    else:
        bot207_message = random.choice(responses['default'])
    return bot207_message
```

```python
def real(text):
  if 'name' in text:
    text_out = "what's your name?"
  elif text=='hello' or text=='hey' :
    text_out = 'hello'
  elif ('weather' in text) or ('climate' in text):
    text_out = ("what's today's weather?")
  elif 'how are' in text:
    text_out = "how are you?"
  elif text=='stop' or text=='exit' or text=='end':
    text_out = "end"
  elif text=='':
    text_out = ""
  else:
    text_out = 'default'
  return text_out
```

```python
def send_message(message):
    response = res(message)
    print(response)
```

```python
while True:
  my_input = input()
  my_input = my_input.lower()
  text = real(my_input)
  send_message(text)
  if my_input == 'exit' or my_input=='end' or my_input=='stop':
    break
```
